```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv('/content/student_performance_exploration_v2 - student_performance_exploration_v2.csv')
```

Histogram-It groups values into ranges and shows how many observations fall in each range.

 Bar chart — It compares categories, such as average scores by department or sales by product type.


Scatter plot — A scatter plot to examine two numerical variables. Each dot represents one observation.


Box plot — Compares a numerical variable across categories. The line inside the box marks the median the box shows the middle half of the data. Points beyond whiskers may be potential outliers, but an outlier is not automatically an error.


```python
sns.histplot(data=df, x="midterm_score")
plt.title("Distribution of Midterm Scores")
plt.xlabel("Midterm score")
plt.ylabel("Number of students")
plt.show()
```

A heatmap is a type of data visualization that uses different colors to represent numerical values.

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="viridis")
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Blues")
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Reds")
```

```python
import numpy as np

corr = df.corr(numeric_only=True)

mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    mask=mask
)

plt.title("Triangle Correlation Heatmap")
plt.show()
```
